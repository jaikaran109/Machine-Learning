### Filtering a DataFrame

```python
ipl.head(2)
```

```python
# find all the final winners
mask = ipl['MatchNumber'] == 'Final'    # yha se tmne jo matches final thi unke fetch kr liya
new_df = ipl[mask]     # isme tmne bss final wale matches ka data nikal liya
new_df[['Season','WinningTeam']]    # yha bss vo matches aayenge jo final thi means winning thi along with its season and note that season phle aayega bcz vo phle pass kiya gya h


#         jo matches final thi        season    winning team
# ipl[ipl['MatchNumber'] == 'Final'][['Season','WinningTeam']]    # same uper wali cheej ko likhne ka shorter syntax
```

```python
# how many super over finishes have occured
ipl[ipl['SuperOver'] == 'Y']   # Y means yes

# ipl[ipl['SuperOver'] == 'Y'].shape[0]   -- kitne number of matches thi , since the entire thing is data of superover matches soo if ill count the number of rows means that is the number of superover matches in whole ipl dataset
```

```python
# how many matches has csk won in kolkata

# ab dekho yha 2 conditions h first city kolkata hona chahiye and winning team csk honi chahiye

#         city filter                   winning team csk honi chahiye
ipl[(ipl['City'] == 'Kolkata') & (ipl['WinningTeam'] == 'Chennai Super Kings')]


# ipl[(ipl['City'] == 'Kolkata') & (ipl['WinningTeam'] == 'Chennai Super Kings')].shape[0]    same agr tm number of matches find krna chahte ho to shape dekh lo
```

```python
# toss winner is match winner in percentage   --  toss jitne se fayda hota h ki nhi although bahut bekar ques h but kya kr skte h video me kraya h

# yha 2 conditions h toss jitna and match jitna --  toss bhi jita and match bhi bss uska percentage nikal do
(ipl[ipl['TossWinner'] == ipl['WinningTeam']].shape[0]/ipl.shape[0])*100
```
