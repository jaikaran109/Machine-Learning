```python
# Import Library

import numpy as np
```

```python
arr1 = np.array([1,2,3,4,5]);
print(arr1);
arr2 = np.arange(0,10,2);  # print all even numbers 0 2 4 8
print(arr2);
arr3 = np.linspace(0,1,5);  # 5 values between 0 to 1
print(arr3);
zeros_arr = np.zeros((2,3));  #create matrix with value 0
print(zeros_arr);
ones_arr = np.ones((2,3));  # create matrix with value 1
print(ones_arr)
full_arr = np.full((3,4),7) # create 3x4 matrix with value 7
print(full_arr)
```

```python
#Random Arrays

rand_arr = np.random.rand(3,3); # Uniform distribution -- $[0, 1)$ — all numbers are between 0 and 1
print(rand_arr)

print("---------------------------------")

randn_arr = np.random.randn(3,3); # Standard Normal distribution (Gaussian) -- (-infinity, +infinity) — typically clustered around 0
print(randn_arr)

print("---------------------------------")

random_arr = np.random.random()  # Single Value
print(random_arr)
```

```python
# Operations
multiplier = np.multiply(arr1,2);  # multiply array1 by 2
print(multiplier)

dot_product = np.dot(arr1,arr1); # Dot Product
print(dot_product)

matrix_mult = np.matmul(rand_arr,randn_arr) # both must have same structure
print(matrix_mult)

product_all = np.prod(arr1)  # product of array
print(product_all)
```

```python
# Properties

print("Shape : " , rand_arr.shape);  # give structure of array
print("Dimention : " , rand_arr.ndim) # give dimensions
print("Size : " , rand_arr.size)  # number of elements
print("Bytes : " , rand_arr.nbytes) # Total Bytes = Number of Elements × Size of One Element -- Each element occupy 8bit , soo output is 72 bytes
```
